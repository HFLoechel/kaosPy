# Basic Chaos Game Representation

<br>
This example demonstrates how to create a Chaos Game Representation (CGR) from a sequence. It covers the basic construction, visualization, and access to the generated coordinates and object properties.

```python
from kaospy import CGR
import matplotlib.pyplot as plt



# Suppress the warning that is emitted when symbols are generated automatically. 
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

<br>

## Constructing a CGR

<br>
Creating a CGR from the sequence "ABRACADABRA" and visualize the resulting point sequence.

```python
text = "ABRACADABRA"

text_cgr = CGR(text)
text_cgr.plot()
plt.show()
```

<br>
Display a short overview of the created CGR object:

```python
text_cgr
```

<br>

## Accessing coordinates

The generated coordinates can be accessed in different ways. The complete coordinate set, the individual x- and y-coordinates, and single coordinates are all available through dedicated methods.

<br>
Get the x- and y-coordinates of the generated CGR:

```python
text_cgr.coordinates
```

<br>
Get the x- coordinates of the generated CGR:

```python
text_cgr.x_coords
```

<br>
Get the y-coordinates of the generated CGR:

```python
text_cgr.y_coords
```

<br>
Get a coordinate of the generated CGR by index:

```python
text_cgr.coordinate_at(0)
```

<br>

## Object properties

<br>
Getting the number of generated coordinates, which is equal to the length of the input sequence:

```python
len(text_cgr)
```

<br>
Getting the radius. The default value is 1 and corresponds to the unit circle:

```python
text_cgr.radius
```

<br>
Symbols can either be generated automatically, selected from one of the predefined alphabets, or provided explicitly:

```python
# Symbols for constructing the CGR, if not specified the symbols will be taken from the input sequnece and there will
#be a warning about it
warnings.filterwarnings("default", category=UserWarning)
text_cgr=CGR(text)
text_cgr.symbols
```

<br>
Symbols can be set with predefined alphabets:

```python
text_cgr=CGR(text, symbols="LETTERS")
text_cgr.symbols
```

<br>
Symbols can also be set manually:

```python
text_cgr=CGR(text, symbols=["A","B","C","D","N","R"])
text_cgr.symbols
```

<br>
Getting the scaling factor:

```python
text_cgr.scaling_factor
```

<br>
Getting the mapping of symbols to vertices:

```python
text_cgr.vertices
```
