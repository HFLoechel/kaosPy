# Basic Frequency Chaos Game Representation

<br>
This example demonstrates how to create a Frequency Chaos Game Representation (FCGR) from a sequence. It covers the basic construction, visualization, and access to the generated coordinates and object properties.

```python
from kaospy import FCGR
from kaospy import CGR

import matplotlib.pyplot as plt



# Suppress the warning that is emitted when symbols are generated automatically. 
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

<br>

## Constructing a FCGR

<br>
Creating a FCGR from the sequence "ABRACADABRA" and visualize the resulting FCGR.

```python
text = "ABRACADABRA"

text_fcgr = FCGR(text, resolution=10)
text_fcgr.plot()
plt.show()
```

<br>
Creating a FCGR from an existing CGR:

```python
text = "ABRACADABRA"

text_cgr = CGR(text)
text_fcgr = FCGR(text_cgr, resolution=10)
text_fcgr.plot()
plt.show()
```

<br>
Display a short overview of the created FCGR object:

```python
text_fcgr
```

<br>

## Accessing coordinates

The generated coordinates can be accessed in the same way as for CGRs:

```python
text_fcgr.coordinates
text_fcgr.x_coords
text_fcgr.y_coords
text_fcgr.coordinate_at(0)
```

<br>

## Object properties

<br>
Since an CGR is needed to create an FCGR the same object properties are availiable:

```python
len(text_fcgr)
text_fcgr.radius
```

<br>
There are some additional properties specific for FCGR:
<br>
The `ignore_boundary` boolean controls how intersections with the grid boundary are handled for FCGR. If set to `False`, points on the boundary are ignored:

```python
text_fcgr.ignore_grid_line_points
```

<br>
Getting the resolution for constructing the matrix:

```python
text_fcgr.resolution
```

<br>
Getting the FCGR matrix:

```python
text_fcgr.matrix
```

<br>
It is possible to set a new resolution:

```python
text_fcgr.set_resolution(50)
text_fcgr.resolution
text_fcgr.matrix
```
