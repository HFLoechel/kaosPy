# N-flakes
<br>
PyKaos can generate (F)CGRs for any number of symbols and scaling factors. For demonstration, we use random numbers.
Random numbers cover the entire space of the area used for the (F)CGR, so sequences from text, DNA, or proteins will look different.

```python
from kaospy import CGR
from kaospy import FCGR
import matplotlib.pyplot as plt
import random


# Suppress the warning that is emitted when symbols are generated automatically. 
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

<br>

## Sierpinski Triangle
<br>
The Sierpinski triangle can be generated using a CGR with three symbols. We create a random sequence of three symbols and visualize both the CGR and the corresponding FCGR.

```python
num_symbols = 3
sequence = [random.randint(1, num_symbols) for _ in range(100000)]
cgr = CGR(sequence)

fcgr = FCGR(cgr,resolution=100)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

cgr.plot(ax=axes[0])
axes[0].set_title("CGR")

fcgr.plot(ax=axes[1])
axes[1].set_title("FCGR")

plt.tight_layout()
plt.show()
```

<br>

## Square (4 Symbols)
<br>
Using four symbols results in a square geometry. By increasing the scaling factor from the default value of 0.5 to 0.6, the generated pattern changes noticeably:

```python
num_symbols = 4
sequence = [random.randint(1, num_symbols) for _ in range(100000)]

cgr_default = CGR(sequence)
cgr_sf_06 = CGR(sequence, scaling_factor=0.6)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

cgr_default.plot(ax=axes[0])
axes[0].set_title("Default scaling factor (0.5)")

cgr_sf_06.plot(ax=axes[1])
axes[1].set_title("Scaling factor = 0.6")

plt.tight_layout()
plt.show()
```

<br>

## General n-flake
The following example generates an n-flake using the default scaling factor. Change the value of `n` to explore how the CGR and FCGR change with the number of symbols.

```python
n=5 # number of vertices
```

```python
sequence = [random.randint(1, n) for _ in range(100000)]

cgr = CGR(sequence)
fcgr = FCGR(cgr)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

cgr.plot(ax=axes[0])
axes[0].set_title("CGR")

fcgr.plot(ax=axes[1])
axes[1].set_title("FCGR")

plt.tight_layout()
plt.show()
```

<br>

## Effect of the scaling factor
<br>
The default scaling factor generates an n-flake without overlapping regions. Reducing the scaling factor produces overlapping structures.

```python
n = 6  # Number of symbols
```

```python
sf =0.5 #Scaling factor
```

```python
sequence = [random.randint(1, n) for _ in range(100000)]

fcgr_default = FCGR(sequence)
fcgr_custom = FCGR(sequence, scaling_factor=sf)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

fcgr_default.plot(ax=axes[0])
axes[0].set_title(
    f"Default scaling factor ({fcgr_default.scaling_factor:.3f})"
)

fcgr_custom.plot(ax=axes[1])
axes[1].set_title(f"Scaling factor = {sf}")

plt.tight_layout()
plt.show()
```
