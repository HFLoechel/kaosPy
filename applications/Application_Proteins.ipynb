# Applications: Proteins
<br>

This notebook demonstrates applications of Chaos Game Representation (CGR)
and Frequency Chaos Game Representation (FCGR) for Proteins.

Examples include:


```python
from kaospy import CGR
from kaospy import FCGR
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import BoundaryNorm
```

<br>

## HIV -V3 Loop

The first example uses an HIV-V3 (1.RU.-.RU29) 

```python

v3="CTRPNNNTRKSIRIGPGQAFYATGEIIGDIRKAYC"
```

<br>
A standard Protein alphabet is used, in alphabetic order and the CGR
coordinates are calculated using the default assignment.
<br>

```python
v3_cgr = CGR(v3, symbols="AMINO") 
v3_fcgr=FCGR(v3_cgr,resolution=10)


fig, axes = plt.subplots(1, 2, figsize=(10, 5))

v3_cgr.plot(ax=axes[0])
axes[0].set_title("CGR")

v3_fcgr.plot(ax=axes[1])
axes[1].set_title("FCGR")

plt.tight_layout()
plt.show()
```

<br>

## Importance of specifying symbols

CGR is based on a mapping between symbols and geometric positions.
If no symbol set is provided, kaosPy derives the symbols directly from the
sequence.

For biological sequences, explicitly defining the symbols is recommended.
For example, protein sequences should be processed with the amino acid
symbol set to ensure the intended biological interpretation.

```python
v3_cgr_amino = CGR(v3, symbols="AMINO")
v3_fcgr_amino = FCGR(v3_cgr_amino, resolution=10)

v3_cgr = CGR(v3)
v3_fcgr = FCGR(v3_cgr, resolution=10)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

v3_fcgr_amino.plot(ax=axes[0])
axes[0].set_title("Explicit amino acid symbols")

v3_fcgr.plot(ax=axes[1])
axes[1].set_title("Symbols inferred from sequence")

plt.tight_layout()
plt.show()
```
